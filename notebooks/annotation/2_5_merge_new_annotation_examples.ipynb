# Imports

```python
import pandas as pd
import seaborn as sns
```

# Constants

```python
HIGH_CONFIDENCE_FILE = "high_confidence_annotation_examples.json"
MEDIUM_CONFIDENCE_FILE = "medium_confidence_annotation_examples.json"
LOW_CONFIDENCE_FILE = "low_confidence_annotation_examples.json"
NO_CONFIDENCE_FILE = "no_confidence_annotation_examples.json"
```

# Load datasets

```python
df_high = pd.read_json(HIGH_CONFIDENCE_FILE, lines=True)
df_medium = pd.read_json(MEDIUM_CONFIDENCE_FILE, lines=True)
df_low = pd.read_json(LOW_CONFIDENCE_FILE, lines=True)
df_no = pd.read_json(NO_CONFIDENCE_FILE, lines=True)
```

```python
len(df_high), len(df_medium), len(df_low), len(df_no)
```

```python
sum([len(df_high), len(df_medium), len(df_low), len(df_no)])
```

# Merge datasets

```python
df_high["confidence_config"] = "high"
df_medium["confidence_config"] = "medium"
df_low["confidence_config"] = "low"
df_no["confidence_config"] = "no"
```

```python
df_new_annot = pd.concat([
    df_high,
    df_medium,
    df_low,
    df_no
])
```

```python
len(df_new_annot)
```

# Visualizing confidence per group

```python
sns.boxplot(
    df_new_annot,
    x="confidence_config",
    y="confidence"
)
```

# Checking no duplicates

```python
len(df_new_annot.drop_duplicates(["sentence_1", "sentence_2", "dataset"]))
```

# Shuffling to avoid labeler bias per confidence config

```python
df_new_annot = df_new_annot.sample(frac=1)
```

# Saving file

```python
df_new_annot.to_json("new_annotation_examples.json", orient="records", lines=True)
```
