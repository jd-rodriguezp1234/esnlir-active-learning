# Imports

```python
import json
import pandas as pd
```

# Constants

```python
LABEL_FILES = ["new_annotation_examples.json"]
```

# Utils

```python
def convert_jsonl_to_json(jsonl_file_path, output_json_file_path):
    data = []
    
    with open(jsonl_file_path, 'r', encoding='utf-8') as jsonl_file:
        for line in jsonl_file:
            js = json.loads(line.strip())
            if js.get('distance')  != 1.0 :
                js['distance'] = 0
            data.append(js)
            
    
    with open(output_json_file_path, 'w', encoding='utf-8') as json_file:
        json.dump(data, json_file, ensure_ascii=False, indent=4)
```

```python
for file in LABEL_FILES:
    output_file = file.replace(".json", "_label_studio.json")
    convert_jsonl_to_json(file, output_file)
```
