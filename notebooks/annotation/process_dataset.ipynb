# Transform from JSONL to JSON ( Dataset -> LabelStudio )

```python
import json

def convert_jsonl_to_json(jsonl_file_path, output_json_file_path):
    data = []
    
    with open(jsonl_file_path, 'r', encoding='utf-8') as jsonl_file:
        for line in jsonl_file:
            js = json.loads(line.strip())
            if js.get('distance')  != 1.0 :
                js['distance'] = 0
            data.append(js)
            
    
    with open(output_json_file_path, 'w', encoding='utf-8') as json_file:
        json.dump(data, json_file, ensure_ascii=False, indent=4)

input_jsonl_path = './base_dataset/test_length_mismatch.json'
output_json_path = './base_dataset/preprocessed/test_length_mismatch.json'

convert_jsonl_to_json(input_jsonl_path, output_json_path)

print(f'Archivo JSON generado')
```

```python
import json

file_path = './documents_labels/proccessed/Labeled_json_first.json'

with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file) 
```

```python
import json

filtered_data = []

for i in data:
    if i['annotations'][0]['result'] != []:
        sentence_1 = i['data']['sentence_1']
        sentence_2 = i['data']['sentence_2']
        connector_type = i['annotations'][0]['result'][0]['value']['choices'][0]

        filtered_data.append({
            "sentence_1": sentence_1,
            "sentence_2": sentence_2,
            "connector_type": connector_type
        })

output_file_path = './documents_labels/es_wikibooks.jsonl'
with open(output_file_path, 'w', encoding='utf-8') as f:
    for entry in filtered_data:
        f.write(json.dumps(entry, ensure_ascii=False) + '\n')

print(f"Archivo JSONL guardado en {output_file_path}")


```

# Label/Studio to Annotation Report

```python
import pandas as pd
dicts = {}
total = 0
total_par = 0
conteo = {}
df_labeled = {}
for element in data:
    temp = (element['annotations'])
    for i in temp:
        df_labeled[emails[i['completed_by']]] = df_labeled.get(emails[i['completed_by']], {})
        if len(i['result']) > 0:
            df_labeled[emails[i['completed_by']]][element['id']] =  i['result'][0]['value']['choices'][0]

        conteo[element['id']] = conteo.get(element['id'], 0)+1
        dicts[emails[i['completed_by']]] =  dicts.get(emails[i['completed_by']],0) +1
        total+=1
    
    total_par +=1
dicts['Total_anotaciones'] = total
dicts['Total_parec_anotados'] = total_par
```
